## ЗАДАНИЕ 1: Базовые графики matplotlib

```python
# Создайте фигуру с 4 subplots (2x2):
# 1. Линейный график функции y = x^2 * sin(x) на интервале [0, 10]
#    Добавьте заголовок, подписи осей, сетку, легенду
# 2. Гистограмма 1000 случайных точек из нормального распределения
#    Настройте количество bins=30, добавьте заголовок
# 3. Scatter plot: 50 точек (x - равномерное распределение, y = 2x + шум)
#    Настройте размер точек, прозрачность, цвет
# 4. Bar chart: сравнение 5 разных моделей по точности
#    accuracy = [0.85, 0.92, 0.88, 0.95, 0.90]
#    Добавьте значения над столбцами

import numpy as np
import matplotlib.pyplot as plt

# Создаем фигуру с 4 subplots (2x2)
### ваш код здесь ###
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# 1. Линейный график y = x^2 * sin(x)
ax = axes[0, 0]
x = np.linspace(0, 10, 100)
y = x**2 * np.sin(x)
### ваш код здесь ###
ax.plot(x, y, label='y = x² * sin(x)', color='blue', linewidth=2)
ax.set_title('Квадрат умноженный на синус')  # Заголовок
ax.set_xlabel('x')  # Подпись оси X
ax.set_ylabel('y')  # Подпись оси Y
ax.grid(True)  # Включаем сетку
ax.legend()

# 2. Гистограмма нормального распределения
ax = axes[0, 1]
data = np.random.normal(0, 1, 1000)
### ваш код здесь ###
ax.hist(data, bins=30, alpha=0.7, color='blue', edgecolor='black')
ax.set_title('Гистограмма нормального распределения')  # Заголовок
ax.set_xlabel('Значение')  # Подпись оси X
ax.set_ylabel('Частота') # Подпись оси Y

# 3. Scatter plot: y = 2x + шум
ax = axes[1, 0]
x = np.random.uniform(0, 10, 50)
y = 2 * x + np.random.normal(0, 1, 50)
### ваш код здесь ###
ax.scatter(x, y, s=50, alpha=0.6, color='green')  # s - размер, alpha - прозрачность
ax.set_title('Точечная диаграмма: y = 2x + шум')  # Заголовок
ax.set_xlabel('x')  # Подпись оси X
ax.set_ylabel('y')  # Подпись оси Y

# 4. Bar chart: точность моделей
ax = axes[1, 1]
models = ['A', 'B', 'C', 'D', 'E']
accuracy = [0.85, 0.92, 0.88, 0.95, 0.90]
### ваш код здесь ###
bars = ax.bar(models, accuracy, color=['blue', 'blue', 'blue', 'blue', 'blue'])
ax.set_title('Точность моделей')  # Заголовок
ax.set_xlabel('Модель')  # Подпись оси X
ax.set_ylabel('Точность')  # Подпись оси Y
ax.set_ylim(0.8, 1.0)  # Устанавливаем диапазон оси Y

# Добавляем значения над столбцами
for bar, acc in zip(bars, accuracy):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 0.005,
            f'{acc:.2f}', ha='center')

plt.tight_layout()
plt.show()
```

## ЗАДАНИЕ 2: Seaborn визуализации

```python
# Используйте датасет tips из seaborn:
# import seaborn as sns
# tips = sns.load_dataset('tips')

# Создайте 3 графика:
# 1. Box plot: total_bill по дням недели (day)
#    Разделите дополнительно по времени (time) с помощью hue
# 2. Violin plot: tip в зависимости от размера компании (size)
# 3. Pair plot: выберите 4 числовых признака, раскрасьте по полу (sex)

import seaborn as sns
import matplotlib.pyplot as plt

# Загружаем датасет
tips = sns.load_dataset('tips')

# Создаем фигуру для графиков
### ваш код здесь ###
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Box plot: total_bill по дням с разделением по времени
# data=tips, x='day', y='total_bill', hue='time'
### ваш код здесь ###
ax = axes[0]  # Берем левую область
sns.boxplot(data=tips,
            x='day',          # Дни недели по горизонтали
            y='total_bill',   # Сумма счета по вертикали
            hue='time',       # Разные цвета для Lunch/Dinner
            ax=ax,           # Рисуем в левом графике
            palette='Set2')  # Красивые пастельные цвета

ax.set_title('Сумма счета по дням недели\n(Разделение по времени)', fontsize=12)
ax.set_xlabel('День недели')
ax.set_ylabel('Сумма счета ($)')
ax.legend(title='Время')  # Подпись для цветов

# 2. Violin plot: tip по размеру компании
# x='size', y='tip'
### ваш код здесь ###
ax = axes[1]  # Берем правую область
sns.violinplot(data=tips,
               x='size',      # Размер компании по горизонтали
               y='tip',       # Чаевые по вертикали
               ax=ax,        # Рисуем в правом графике
               palette='viridis')  # Радужные цвета

ax.set_title('Чаевые в зависимости от размера компании', fontsize=12)
ax.set_xlabel('Размер компании (чел.)')
ax.set_ylabel('Чаевые ($)')

plt.tight_layout()
plt.show()

# 3. Pair plot: 4 числовых признака, раскраска по полу
# tips[['total_bill', 'tip', 'size', 'sex']]
# Создаем отдельную фигуру для pair plot
### ваш код здесь ###
selected_columns = tips[['total_bill', 'tip', 'size', 'sex']]

# Создаем отдельную фигуру для матрицы графиков
pair_plot = sns.pairplot(selected_columns,
                         hue='sex',      # Раскрашиваем по полу
                         palette='husl', # Яркие цвета
                         height=2.5,     # Высота каждого маленького графика
                         diag_kind='kde')  # Диагональ - плавные кривые вместо гистограмм

pair_plot.fig.suptitle('Взаимосвязь признаков с разбивкой по полу',
                       y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```
